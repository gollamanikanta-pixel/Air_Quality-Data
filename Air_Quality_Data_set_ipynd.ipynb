```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Air_Quality.csv')

# Clean and split the combined column into PM2.5 (numeric) and Category (text)
df[['PM2_5', 'Air_Quality_Category']] = df['Unnamed: 3'].str.split(' ', n=1, expand=True)
df['PM2_5'] = pd.to_numeric(df['PM2_5'])

# --- DATA PREPARATION & AGGREGATION ---
# Filter data for just the City location
city_center_data = df[df['Location'] == 'City']

# Group data by location to find average pollution levels
avg_pollution = df.groupby('Location')[['PM2_5']].mean()

# --- THE 5 REQUIRED CHARTS ---
# Chart 1: Line Chart (Tracking PM2.5 over time at City)
plt.figure(figsize=(8, 4))
plt.plot(city_center_data['Date'],
         city_center_data['PM2_5'],
         marker='o', color='red')
plt.title('Chart 1: PM2.5 Trend at City Center')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by Location)
plt.figure(figsize=(8, 4))
avg_pollution['PM2_5'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 levels over Date)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2_5'], color='purple', s=140)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5 Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the dataset
df = pd.read_csv('Air_Quality.csv')

# 2. Clean and split the combined column into numeric PM2.5 and Category
df[['PM2_5', 'Air_Quality_Category']] = df['Unnamed: 3'].str.split(' ', n=1, expand=True)
df['PM2_5'] = pd.to_numeric(df['PM2_5'])

# 3. Chart 4: Distribution of PM2.5 Levels (Histogram)
plt.figure(figsize=(8, 4))
plt.hist(df['PM2_5'], bins=5, color='skyblue', edgecolor='black')
plt.title('Chart 4: Distribution of PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Chart 5: Horizontal Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2_5'].max().plot(kind='barh', color='green')
plt.title('Chart 5: Max PM2.5 Levels by Location')
plt.xlabel('Max PM2.5 Level')
plt.ylabel('Location')
plt.show()
```

```python
# Chart 5: Vertical Bar Chart (Max PM2.5 recorded per location)
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2_5'].max().plot(kind='bar', color='blue')
plt.title('Chart 5: Max PM2.5 Recorded per Location')
plt.xlabel('Location')
plt.ylabel('Max PM2.5')
plt.xticks(rotation=0)
plt.show()
```
